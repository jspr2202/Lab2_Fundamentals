```python
# EXERCISE 1 - SENSOR READING
import random
random.seed(SEED_NUM)
sensor_data = [
    random.randint(0, 100)
    for _ in range(10)
]
print("Generated Sensor Data:", sensor_data)
```

```python
# Validate sensor readings
valid_results = []
for reading in sensor_data:
    if reading in sensor_data:
        if isinstance(reading, (int, float)) and 0 <= 100:
            valid_results.append((reading, "Valid"))
    else:
        valid_results.append((reading, "Invalid"))
print("Valid/Invalid Results:")
for reading, status in valid_results:
    print(f"Reading: {reading} | Status: {status}")
```

```python
# Classify valid sensor readings
classified_results = []
for reading, status in valid_results:
    if status == "Valid":
        if 0 <= reading < 30:
            classification = "Low"
        elif 30 <= reading <= 70:
            classification = "Normal"
        else:
            classification = "High"
        classified_results.append((reading, classification))
print("Classification Results:")
for reading, classification in classified_results:
    print(f"Reading: {reading} | Classification: {classification}")
```

```python
# Exercise 1 Summary
valid_count = sum(1 for _, status in valid_results if status == "Valid")
invalid_count = sum(1 for _, status in valid_results if status == "Invalid")
print("=========== SENSOR SUMMARY ===========")
print(f"Generated Sensor Data: {sensor_data}")
print(f"Valid Readings: {valid_count}")
print(f"Invalid Readings: {invalid_count}")
print("Classification of Valid Readings:")
for reading, classification in classified_results:
    print(f"Reading: {reading} | Classification: {classification}")
print("======================================")
```
